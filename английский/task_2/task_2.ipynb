## Импорты

```python
from PIL import Image
import numpy as np
```

## Получение numpy массива из изборажения

```python
IMG_NAME = "image.png"
img = Image.open(IMG_NAME).convert("RGB")
arr = np.asarray(img)
arr.shape
```

## Обработка изображения

```python
# получаем маску массива с максимумами
mask = (arr == arr.max(axis=2, keepdims=True)).astype(np.uint8)

# изменяем изображение
processed_arr = arr * mask
```

## Сохранение изображения

```python
processed_img = Image.fromarray(processed_arr)
processed_img.save(f'processed_{IMG_NAME}')
```
